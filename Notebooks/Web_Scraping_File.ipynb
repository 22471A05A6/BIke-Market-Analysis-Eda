```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

all_names = []
all_specs = []
all_price = []
all_ratings = []
No_of_ratings = []
No_of_reviews = []

for page in range(1, 30):

    url = f"https://www.flipkart.com/search?q=bikes&page={page}"

    response = requests.get(url, headers=headers)

    print(page, response.status_code)

    soup = BeautifulSoup(response.text, "html.parser")

    # Product Name
    names = soup.find_all("div", class_="RG5Slk")

    # Specifications
    specs = soup.find_all("div", class_="CMXw7N")

    # price 
    price = soup.find_all("div",class_="hZ3P6w DeU9vF")

    # Rating 
    rating = soup.find_all("div",class_="MKiFS6")

    # No of Ratings
    ratings_no = soup.find_all("span",class_="PvbNMB")

    # No of Reviews
    reviews_no = soup.find_all("span",class_="PvbNMB")

    for n in names:
        all_names.append(n.text.strip())

    for s in specs:
        all_specs.append(s.text.strip())
    for p in price:
        all_price.append(p.text.strip())
    for r in rating:
        all_ratings.append(r.text.strip())
    for nr in ratings_no:
        No_of_ratings.append(nr.text.strip())
    for nre in reviews_no:
        No_of_reviews.append(nre.text.strip())
```

```python
print(len(all_names))
print(len(all_specs))
print(len(all_price))
print(len(all_ratings))
print(len(No_of_reviews))
print(len(No_of_ratings))
```

```python
min_len=min(len(all_names),
            len(all_specs),
            len(all_price),
            len(all_ratings),
            len(No_of_ratings),
            len(No_of_reviews))
min_len
```

```python
all_names = all_names[:min_len]
all_specs = all_specs[:min_len]
all_price = all_price[:min_len]
all_ratings = all_ratings[:min_len]
No_of_ratings = No_of_ratings[:min_len]
No_of_reviews = No_of_reviews[:min_len]
```

```python
import pandas as pd

df = pd.DataFrame({
    "Product_Name": all_names,
    "Specifications": all_specs,
    "Price": all_price,
    "Rating": all_ratings,
    "No_of_Ratings": No_of_ratings,
    "No_of_Reviews": No_of_reviews
})
```

```python
brands = [
    "Hero","Honda","TVS","Yamaha","Bajaj",
    "Royal Enfield","KTM","Suzuki","Ather",
    "VIDA","Yezdi","Chetak"
]

def get_brand(name):
    for brand in brands:
        if brand.lower() in name.lower():
            return brand
    return "Other"

df["Brand"] = df["Product_Name"].apply(get_brand)
```

```python
def bike_type(name):
    name = name.lower()

    if "electric" in name:
        return "Electric"

    elif "scooter" in name:
        return "Scooter"

    else:
        return "Motorcycle"

df["Bike_Type"] = df["Product_Name"].apply(bike_type)
```

```python
def clean_product_name(row):
    name = row["Product_Name"]
    brand = row["Brand"]

    if pd.notna(name) and pd.notna(brand):
        if name.lower().startswith(brand.lower()):
            name = name[len(brand):].strip()

    return name

df["Product_Name"] = df.apply(clean_product_name, axis=1)
```

```python
def split_tire(text):
    if pd.isna(text):
        return pd.Series([None, None])
    text = str(text)
    for tire in ['Tubeless', 'Tubed']:
        if text.startswith(tire):
            return pd.Series([tire, text[len(tire):].strip() or None])
    return pd.Series([None, text])

df[['Tire_Type_clean', 'Extra_Desc_1']] = df['Tire_Type'].apply(split_tire)
```

```python
def convert_to_bhp(power):

    if pd.isna(power):
        return np.nan

    text = str(power).lower()

    # ----------- BHP -----------
    bhp = re.search(r'(\d+\.?\d*)\s*bhp', text)
    if bhp:
        return float(bhp.group(1))

    # ----------- PS -----------
    ps = re.search(r'(\d+\.?\d*)\s*ps', text)
    if ps:
        return round(float(ps.group(1)) * 0.98632, 2)

    # ----------- kW -----------
    kw = re.search(r'(\d+\.?\d*)\s*kw', text)
    if kw:
        return round(float(kw.group(1)) * 1.34102, 2)

    # ----------- Only Number -----------
    num = re.search(r'(\d+\.?\d*)', text)
    if num:
        return float(num.group(1))

    return np.nan
```

```python
df_non_electric["Maximum_Power_BHP"] = (
    df_non_electric["Maximum_Power"]
    .apply(convert_to_bhp)
)
```
