# Bivariate Analysis

#### Price vs Rating
* Do expensive bikes receive higher ratings?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Price_INR",
    y="Rating",
    hue="Brand_fixed"
)
plt.title("Price vs Rating By Rating")
plt.xlabel("Price (INR)")
plt.ylabel("Rating")

plt.show()
```

### Price vs Engine Capacity
* Does engine capacity affect bike price?

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Engine_Capacity_CC",
    y="Price_INR",
    hue="Brand_fixed"
)

plt.title("Engine Capacity vs Price")

plt.show()
```

### Price vs Maximum Power
Do powerful bikes cost more?

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Maximum_Power_BHP",
    y="Price_INR"
)

plt.title("Maximum Power vs Price")

plt.show()
```

### Brand vs Price
Which brands sell the most expensive bikes?

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Brand_fixed",
    y="Price_INR",
    estimator="mean",
    errorbar=None
)

plt.xticks(rotation=45)

plt.title("Average Price by Brand")

plt.show()
```

### Brand vs Maximum Power
Which brand offers the most powerful bikes?

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Brand_fixed",
    y="Maximum_Power_BHP",
    estimator="mean",
    errorbar=None
)

plt.xticks(rotation=45)

plt.title("Average Power by Brand")

plt.show()
```

### Front Brake vs Price
*Are disc brake bikes more expensive?

```python
plt.figure(figsize=(7,5))

sns.histplot(
    data=df,
    x="Front_Brake",
    y="Price_INR"
)

plt.title("Front Brake vs Price")

plt.show()
```
