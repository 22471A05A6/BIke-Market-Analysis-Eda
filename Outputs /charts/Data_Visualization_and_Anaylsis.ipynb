## Brand-wise Summary

```python
brand_summary = df.groupby("Brand_fixed").agg(
    Average_Price=("Price_INR","mean"),
    Average_Rating=("Rating","mean"),
    Average_Power=("Maximum_Power_BHP","mean"),
    Bike_Count=("Product_Name","count")
).round(2)

brand_summary
```

### Brand-wise Ranking

```python
brand_summary.sort_values(
    by="Average_Price",
    ascending=True
)
```

### Engine Capacity Analysis

```python
engine_summary = df.groupby("Engine_Capacity_CC").agg(
    Average_Price=("Price_INR","mean"),
    Bike_Count=("Product_Name","count")
).round(2)

engine_summary
```

### Price Category

```python
bins = [0,100000,200000,400000,1000000]

labels = [
    "Budget",
    "Mid Range",
    "Premium",
    "Luxury"
]

df["Price_Category"] = pd.cut(
    df["Price_INR"],
    bins=bins,
    labels=labels
)
```

```python
df["Price_Category"].value_counts()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Price_Category"
)

plt.title("Bike Price Categories")

plt.show()
```

### Business Recommendations
14. Expand the Mid-Range Segment

Manufacturers should continue introducing more motorcycles in the ₹–₹ price range, as this segment represents the largest share of customer demand.

15. Focus on Performance-Oriented Models

Brands should continue developing motorcycles with higher engine capacity and better power output, as these features strongly influence customer purchase decisions.

16. Improve Customer Experience

Brands with comparatively lower customer ratings should improve build quality, riding comfort, fuel efficiency, after-sales service, and reliability to enhance customer satisfaction.

17. Increase Safety Features

Manufacturers should make disc brakes, ABS, and tubeless tires standard features across more motorcycle models to improve rider safety.

18. Introduce More Technology Features

Digital instrument clusters, Bluetooth connectivity, navigation, and smartphone integration are increasingly preferred by customers. Expanding these features can improve product competitiveness.

19. Maintain Competitive Pricing

Brands should regularly compare prices with competitors to ensure customers receive better value while maintaining product quality and profitability.

20. Promote High-Rated Bikes

Motorcycles with excellent customer ratings should receive greater visibility through online promotions, featured listings, and marketing campaigns to increase sales.

21. Strengthen Brand Portfolio

Brands with fewer motorcycle models should expand their product lineup across commuter, sports, cruiser, and adventure categories to improve market presence and attract a wider customer base.
